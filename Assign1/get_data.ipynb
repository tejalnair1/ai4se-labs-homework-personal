```python
df = pd.read_csv("data/python_repos.csv")
```

```python
# Build URLs directly from name column
repo_urls = ["https://github.com/" + n for n in df["name"].dropna().unique()]
```

```python
with open("data/repos_list.txt", "w") as f:
    for url in repo_urls:
        f.write(url + "\n")
```
